```python
import pandas as pd
df=pd.read_csv("Amazon Sale Report.csv")
df.head()
```

```python
delivery_statuses = [
    "Shipped - Delivered to Buyer",
    "Shipped - Returned to Seller",
    "Shipped - Returning to Seller",
    "Shipped - Rejected by Buyer",
    "Shipped - Lost in Transit",
    "Shipped - Damaged",
    "Shipped - Out for Delivery",
    "Shipped - Picked Up"
]

delivery_outcomes = df[df["Status"].isin(delivery_statuses)]

print("Delivery & Return Outcome Distribution:\n")
print(delivery_outcomes["Status"].value_counts())
```

```python
total_delivery_records=len(delivery_outcomes)

delivered_records = (
    delivery_outcomes["Status"] == "Shipped - Delivered to Buyer"
).sum()

delivery_success_rate = (
    delivered_records / total_delivery_records
) * 100

print("Total Delivery-related Records:", total_delivery_records)
print("Delivered Records:", delivered_records)
print("Delivery Success Rate:", round(delivery_success_rate, 2), "%")
```

```python
return_statuses = [
    "Shipped - Returned to Seller",
    "Shipped - Returning to Seller"
]

returned_records = delivery_outcomes[
    delivery_outcomes["Status"].isin(return_statuses)
].shape[0]

return_rate = (
    returned_records / total_delivery_records
) * 100

print("Returned/Returning Records:", returned_records)
print("Return Rate:", round(return_rate, 2), "%")
```

```python
problem_statuses = [
    "Shipped - Rejected by Buyer",
    "Shipped - Lost in Transit",
    "Shipped - Damaged"
]

problem_records = delivery_outcomes[
    delivery_outcomes["Status"].isin(problem_statuses)
].shape[0]

problem_rate = (
    problem_records / total_delivery_records
) * 100

print("Problematic Delivery Records:", problem_records)
print("Problematic Delivery Rate:", round(problem_rate, 2), "%")
```

```python
in_progress_statuses = [
    "Shipped - Picked Up",
    "Shipped - Out for Delivery"
]

in_progress_records = delivery_outcomes[
    delivery_outcomes["Status"].isin(in_progress_statuses)
].shape[0]

in_progress_rate = (
    in_progress_records / total_delivery_records
) * 100

print("In-Progress Delivery Records:", in_progress_records)
print("In-Progress Delivery Rate:", round(in_progress_rate, 2), "%")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Amazon Sale Report.csv")

revenue = pd.pivot_table(
    df,
    index="Category",
    values="Amount",
    aggfunc="sum"
).sort_values(by="Amount", ascending=False)

print(revenue)

# Visualization of Revenue Generated by Product Category
plt.figure(figsize=(10, 6))

revenue_millions = revenue["Amount"] / 1_000_000

bars = plt.bar(
    revenue.index,
    revenue_millions,
    color="orange"
)

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue (₹ Million)")

plt.xticks(rotation=45)

for bar in bars:
    value = bar.get_height()
    
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"₹{value:.1f}M",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

# Save visualization for GitHub
plt.savefig(
    "visualizations/01_revenue_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

print(df["Date"].dtype)


# Create Month-Year column
df["Month-Year"] = df["Date"].dt.to_period("M")

print(df[["Date", "Month-Year"]].head())


# Calculate Monthly Revenue
monthly_revenue = (
    df.groupby("Month-Year")["Amount"]
      .sum()
      .sort_index()
)

print("Monthly Revenue:\n")
print(monthly_revenue)


# Keep only full comparable months
revenue_full_months = monthly_revenue.loc["2022-04":"2022-06"]


# Visualization
plt.figure(figsize=(10, 6))

plt.plot(
    revenue_full_months.index.astype(str),
    revenue_full_months.values / 1_000_000,
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹ Million)")

plt.grid(True)

plt.tight_layout()

# Save Visualization 5
plt.savefig(
    "visualizations/05_monthly_revenue_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Monthly Quantity Trend

# Calculate monthly quantity sold
monthly_quantity = (
    df.groupby("Month-Year")["Qty"]
      .sum()
      .sort_index()
)

print("Monthly Quantity:\n")
print(monthly_quantity)


# Keep only full comparable months
quantity_full_months = monthly_quantity.loc["2022-04":"2022-06"]


# Visualization
plt.figure(figsize=(10, 6))

plt.plot(
    quantity_full_months.index.astype(str),
    quantity_full_months.values / 1_000,
    marker="o"
)

plt.title("Monthly Quantity Sold Trend")
plt.xlabel("Month")
plt.ylabel("Quantity Sold (Thousands)")

plt.grid(True)

plt.tight_layout()

# Save Visualization 6
plt.savefig(
    "visualizations/06_monthly_quantity_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Monthly Quantity
monthly_quantity = (
    df.groupby("Month-Year")["Qty"]
      .sum()
      .sort_index()
)

# Monthly Average Selling Price (ASP)
monthly_asp = (
    monthly_revenue /
    monthly_quantity
)

print("Monthly Average Selling Price (ASP):\n")
print(monthly_asp.sort_index())


# Keep only full comparable months
asp_full_months = monthly_asp.loc["2022-04":"2022-06"]


# Visualization
plt.figure(figsize=(10, 6))

plt.plot(
    asp_full_months.index.astype(str),
    asp_full_months.values,
    marker="o"
)

plt.title("Monthly Average Selling Price (ASP) Trend")
plt.xlabel("Month")
plt.ylabel("Average Selling Price (₹)")

plt.grid(True)

plt.tight_layout()

# Save visualization
plt.savefig(
    "visualizations/08_monthly_asp_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Category-wise Average Selling Price (ASP)

revenue = pd.pivot_table(
    df,
    index="Category",
    values="Amount",
    aggfunc="sum"
)

Qty = pd.pivot_table(
    df,
    index="Category",
    values="Qty",
    aggfunc="sum"
)

# Calculate and sort ASP
Avg_Selling_price = (
    revenue["Amount"] / Qty["Qty"]
).sort_values(ascending=False)

print("Average Selling Price:\n")
print(Avg_Selling_price)


# Visualization
plt.figure(figsize=(10, 6))

bars = plt.barh(
    Avg_Selling_price.index[::-1],
    Avg_Selling_price.values[::-1]
)

plt.title("Average Selling Price by Product Category")
plt.xlabel("Average Selling Price (₹)")
plt.ylabel("Product Category")

for bar in bars:
    value = bar.get_width()

    plt.text(
        value,
        bar.get_y() + bar.get_height() / 2,
        f"₹{value:.0f}",
        ha="left",
        va="center"
    )

plt.tight_layout()

# Save the visualization
plt.savefig(
    "visualizations/07_average_selling_price_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
